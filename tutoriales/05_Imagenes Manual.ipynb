```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, cohen_kappa_score
import os

import cv2
import matplotlib.pyplot as plt

from utils import plot_confusion_matrix
```

**Seteo parámetros, directorios y funciones**

```python
# Paths
BASE_DIR = '../'
PATH_TO_TRAIN = os.path.join(BASE_DIR, "input/petfinder-adoption-prediction/train/train.csv")
PATH_TO_IMAGES_DIR = os.path.join(BASE_DIR, "input/petfinder-adoption-prediction/train_images")
PATH_TO_TEMP_FILES = os.path.join(BASE_DIR, "work/optuna_temp_artifacts")
PATH_TO_OPTUNA_ARTIFACTS = os.path.join(BASE_DIR, "work/optuna_artifacts")


SEED = 42
BATCH_SIZE = 50
TEST_SIZE = 0.2

# Armo el nuevo directorio de train
new_train_directory = os.path.join(BASE_DIR, 'work/train_images_classes')
os.makedirs(new_train_directory, exist_ok=True) # si ya existe el nombre, lo deja como está

# Armo el nuevo directorio de validación
new_val_directory = os.path.join(BASE_DIR, 'work/val_images_classes')
os.makedirs(new_val_directory, exist_ok=True)

# Definir las clases ordenadas
class_names = ['0', '1', '2', '3', '4']

# Mapear las etiquetas de las clases a números enteros consecutivos
class_to_idx = {class_name: i for i, class_name in enumerate(class_names)}

```

```python

def visualize_pet(pet_id):
    path_to_image = os.path.join(PATH_TO_IMAGES_DIR, f'{pet_id}-1.jpg') # Irá a la primera imagen de la mascota
    # Cargar la imagen
    image_to_show = cv2.imread(path_to_image)
    # Convertir a formato RGB
    image_to_show = cv2.cvtColor(image_to_show, cv2.COLOR_BGR2RGB)
    # Visualizar la imagen
    plt.imshow(image_to_show)
    plt.axis('off')  # No mostrar los ejes
    plt.show()

def visualize_image(image):
    # Convierte la imagen a un formato de enteros (CV_8U)
    image = cv2.convertScaleAbs(image)
    image= cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    # Visualizar la imagen
    plt.imshow(image.astype(np.uint8))
    plt.axis('off')  # No mostrar los ejes
    plt.show()
```

**Cargo y Proceso Data**

Nota: Pytorch necesita que estén las imágenes en los distintos directorios según su clase y su participación en el training

```python
# Cargo
train_df = pd.read_csv(PATH_TO_TRAIN)

# Split para validación
train_data, val_data = train_test_split(train_df,
                               test_size = TEST_SIZE,
                               random_state = SEED,
                               stratify = train_df.AdoptionSpeed)


```

```python
# Organizar imágenes de validación en directorios por clase
import shutil

# Verificar y crear el directorio principal de validación si no existe
if not os.path.exists(new_val_directory):
    os.makedirs(new_val_directory, exist_ok=True)

# Crear directorios para cada clase en validación
for class_name in class_names:
    class_dir = os.path.join(new_val_directory, class_name)
    if not os.path.exists(class_dir):
        os.makedirs(class_dir, exist_ok=True)

# Copiar imágenes de validación a sus respectivos directorios de clase
copied_count = 0
missing_count = 0

for idx, row in val_data.iterrows():
    pet_id = row['PetID']
    adoption_speed = str(row['AdoptionSpeed'])
    
    # Ruta de imagen original
    src_path = os.path.join(PATH_TO_IMAGES_DIR, f'{pet_id}-1.jpg')
    
    # Ruta de destino
    dst_path = os.path.join(new_val_directory, adoption_speed, f'{pet_id}-1.jpg')
    
    # Verificar y crear el subdirectorio de clase si no existe
    class_subdir = os.path.join(new_val_directory, adoption_speed)
    if not os.path.exists(class_subdir):
        os.makedirs(class_subdir, exist_ok=True)

    if os.path.exists(src_path):
        # Copiar imagen al directorio de clase correspondiente
        shutil.copy2(src_path, dst_path)
        copied_count += 1
    else:
        print(f"Imagen no encontrada: {src_path}")
        missing_count += 1

print(f"Imágenes copiadas: {copied_count}")
print(f"Imágenes faltantes: {missing_count}")
print("Organización de imágenes de validación completada.")
```

```python
# Add prediction column
val_data['pred'] = -1

# Loop through dataset
for idx, row in val_data.iterrows():
    # Display image
    img_path = f"../work/val_images_classes/{row['AdoptionSpeed']}/{row['PetID']}-1.jpg"
    
    # Check if image file exists
    if not os.path.exists(img_path):
        print(f"Image not found: {img_path}")
        # Try original images directory instead
        img_path = f"../input/petfinder-adoption-prediction/train_images/{row['PetID']}-1.jpg"
        if not os.path.exists(img_path):
            print(f"Image also not found in original directory: {img_path}")
            continue
    
    img = cv2.imread(img_path)
    
    # Check if image was loaded successfully
    if img is None:
        print(f"Could not load image: {img_path}")
        continue
        
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(8,8))
    plt.imshow(img)
    plt.axis('off')
    plt.show()
    
    # Get user input
    try:
        pred = int(input(f"Enter adoption speed prediction (0-4) for pet {row['PetID']}, or press Ctrl+C to stop: "))
        if pred not in range(5):
            print("\nStopping annotation...")
            break
        val_data.loc[idx, 'pred'] = pred
    except:
        print("\nStopping annotation...")
        break

mask = val_data['pred'] != -1
if mask.any():
    # Calculate and display Cohen's Kappa
    kappa = cohen_kappa_score(val_data.loc[mask, 'AdoptionSpeed'], 
                            val_data.loc[mask, 'pred'],
                            weights='quadratic')
    print(f"\nQuadratic Cohen's Kappa Score: {kappa:.3f}")
    
    # Plot confusion matrix
    display(plot_confusion_matrix(val_data.loc[mask, 'AdoptionSpeed'],
                       val_data.loc[mask, 'pred']))
else:
    print("\nNo predictions were made")
```

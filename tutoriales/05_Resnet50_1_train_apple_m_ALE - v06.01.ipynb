```python
# DirectoryManager v6.01 - Sistema de gestión de directorios inteligente
# Fundación para el pipeline ResNet50 Enhanced

import os
import shutil
import logging
from pathlib import Path
from datetime import datetime
from typing import Dict, List, Optional, Union
import json
import time
from dataclasses import dataclass, asdict

print("🚀 RESNET50 ENHANCED PIPELINE v6.01")
print("📁 Directory Manager - Sistema de gestión de directorios inteligente")
print("=" * 80)

@dataclass
class DirectoryConfig:
    """Configuración de directorios del proyecto."""
    project_root: str = ""
    data_dir: str = "data"
    models_dir: str = "models"
    logs_dir: str = "logs"
    output_dir: str = "output"
    temp_dir: str = "temp"
    backup_dir: str = "backup"

class DirectoryManager:
    """
    Gestor inteligente de directorios para el pipeline ResNet50 Enhanced.
    Maneja la creación, organización y limpieza de toda la estructura de carpetas.
    """
    
    def __init__(self, project_root: Union[str, Path] = None, auto_create: bool = True):
        if project_root is None:
            self.project_root = Path.cwd()
        else:
            self.project_root = Path(project_root)
        
        self.config = DirectoryConfig(project_root=str(self.project_root))
        self.logger = self._setup_logger()
        
        self.stats = {
            "directories_created": 0,
            "files_cleaned": 0,
            "creation_time": datetime.now().strftime('%Y-%m-%d %H:%M:%S')
        }
        
        if auto_create:
            self.setup_directories()
            
        self.logger.info(f"✅ DirectoryManager v6.01 inicializado en: {self.project_root}")
    
    def _setup_logger(self) -> logging.Logger:
        logger = logging.getLogger("DirectoryManager")
        logger.setLevel(logging.INFO)
        
        if not logger.handlers:
            console_handler = logging.StreamHandler()
            console_formatter = logging.Formatter(
                '%(asctime)s - %(name)s - %(levelname)s - %(message)s'
            )
            console_handler.setFormatter(console_formatter)
            logger.addHandler(console_handler)
        
        return logger
    
    def setup_directories(self) -> Dict[str, bool]:
        self.logger.info("📁 Configurando estructura de directorios...")
        
        directories_to_create = [
            "data", "data/raw", "data/processed", "data/augmented",
            "models", "models/checkpoints", "models/final", "models/experiments",
            "logs", "logs/training", "logs/validation", "logs/system",
            "output", "output/predictions", "output/metrics", "output/plots", "output/reports",
            "temp", "backup"
        ]
        
        creation_status = {}
        created_count = 0
        
        for dir_path in directories_to_create:
            full_path = self.project_root / dir_path
            
            try:
                full_path.mkdir(parents=True, exist_ok=True)
                creation_status[dir_path] = True
                created_count += 1
                
            except Exception as e:
                creation_status[dir_path] = False
                self.logger.error(f"❌ Error creando {dir_path}: {e}")
        
        self.stats["directories_created"] = created_count
        self.logger.info(f"✅ Estructura creada: {created_count}/{len(directories_to_create)} directorios")
        
        return creation_status
    
    def get_directory_info(self) -> Dict[str, any]:
        info = {
            "project_root": str(self.project_root),
            "directories": {},
            "total_size_mb": 0,
            "total_files": 0,
            "stats": self.stats
        }
        
        main_dirs = ["data", "models", "logs", "output", "temp"]
        
        for dir_name in main_dirs:
            dir_path = self.project_root / dir_name
            
            if dir_path.exists():
                file_count = len([f for f in dir_path.rglob("*") if f.is_file()])
                subdirs = [d.name for d in dir_path.iterdir() if d.is_dir()]
                
                info["directories"][dir_name] = {
                    "exists": True,
                    "size_mb": 0.0,
                    "file_count": file_count,
                    "subdirectories": sorted(subdirs)
                }
                info["total_files"] += file_count
            else:
                info["directories"][dir_name] = {
                    "exists": False,
                    "size_mb": 0,
                    "file_count": 0
                }
        
        return info

print("✅ DirectoryManager v6.01 definido correctamente")
```
